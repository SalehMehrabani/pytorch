# Building a model with pytorch

```python
class TinyModel(nn.Module):
    def __init__(self):
        super(TinyModel, self).__init__()

        self.linear1= nn.Linear(100, 200)
        self.activation= nn.ReLU()
        self.linear2= nn.Linear(200,10)
        self.sofmax= nn.Softmax()

    def forward(self, x):
        x= self.linear1(x)
        x= self.activation(x)
        x= self.linear2(x)
        x= self.sofmax(x)
        return x

tinymodel=TinyModel()

print("the model:\n", tinymodel)
print("\nthe layer of the model:\n", tinymodel.linear2)
print("\nparameters of the model:")
for param in tinymodel.parameters():
    print(param)

print("\nparameter of the layer 2 of model:")
for param in tinymodel.linear2.parameters():
    print(param)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
# Convolutional Layer
class Lenet(nn.Module):
    def __init__(self ):
        super(Lenet, self).__init__()

        # Input Channel, Output_channel, kernel size
        self.conv1= nn.Conv2d(1,6,5)
        self.conv2= nn.Conv2d(6,16,3)


        self.fc1= nn.Linear(16*6*6, 120)
        self.fc2= nn.Linear(120,84)
        self.fc3= nn.Linear(84,10)

    def num_flat_features(self, x):
        size= x.size()[1:]
        num_features=1
        for s in size:
            num_features*=s
        return num_features

    def forward(self, x):
        # Max pooling over 2x2 window
        x= F.max_pool2d(F.relu(self.conv1(x)), (2,2))

        x= F.max_pool2d(F.relu(self.conv2(x)), 2)

        x= x.view(-1, self.num_flat_features(x))
        x=F.relu(self.fc1(x))
        x= F.relu(self.fc2(x))


        x= self.fc3(x)
        return x
```

```python
# Recurrent layer
class LSTMagger(nn.Module):
    def __init__(self, embedding_dim, hidden_dim, vocab_size, tagset_size):
        super(LSTMagger, self).__init__()

        self.hidden_dim= hidden_dim

        self.word_embedding= nn.Embedding(vocab_size, embedding_dim)
        
        self.lstm= nn.LSTM(embedding_dim, hidden_dim)

        self.hidden_2tag= nn.Linear(hidden_dim, tagset_size)

    
    def forward(self, sentense):
        embeds= self.word_embedding(sentense)
        lstm_out= self.lstm(embeds.view(len(sentense),1,-1))
        tag_space= self.hidden_2tag(lstm_out.view(len(sentense),-1))
        tag_scores= F.log_softmax(tag_space, dim=1)
        return tag_scores

# https://docs.pytorch.org/tutorials/beginner/nlp/sequence_models_tutorial.html
```
