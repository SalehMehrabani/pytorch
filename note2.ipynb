```python
# Common functions
a= torch.rand(2,4)*2-1

print(torch.abs(a))
print(torch.ceil(a))
print(torch.floor(a))
print(torch.clamp(a, -0.5,0.5))
```

```python
# trigonometric functions
angle= torch.tensor([0, math.pi/4, math.pi/2, 3*math.pi/4])
print(angle)
sins= torch.sin(angle)
print(sins)
inversins= torch.arcsin(angle)
print(inversins)
```

```python
# bitwise operations
a= torch.tensor([1,5,11])
b= torch.tensor([2, 7, 10])
print(torch.bitwise_xor(a,b))
```

```python
# comparisons
a= torch.tensor([[1.,2.],[3.,4.]])
b= torch.ones(1,2)

print(torch.eq(a,b))
```

```python
# reductions
a= torch.rand(3,2)

print(torch.max(a))
print(torch.max(a).item()) # return the item 
print(torch.mean(a))
print(torch.std(a))
print(torch.prod(a))

a= torch.tensor([1,2,3,4,1,3,2,4])
print(torch.unique(a))
```

```python
# Vector and linear algebra

v1= torch.tensor([1,2,1])
v2= torch.tensor([2,1,3])
m1= torch.rand(2,2)
m2=torch.tensor([[1,3],[3,1]]).to(torch.float32)
print(m2.dtype)
m3=torch.matmul(m1,m2)

print(torch.cross(v2,v1))
print(m3)
print(torch.svd(m3))
```

```python
angle= torch.tensor([0, math.pi/4, math.pi/2, 3*math.pi/4])

print(torch.sin(angle))# the initial value of angle remain the same
print(angle)
print(torch.sin_(angle))# the inplace operation the, value of the angle tensor changed
print(angle)

a= torch.rand(3,2)
b= torch.ones(3,2)
print(b)
print(a)
print(b.add_(a))
print(b)
print(a)
```

```python
a= torch.rand(2,2)
b= torch.rand(2,2)
c= torch.zeros(2,2)
old_id= id(c)

print(c)
d= torch.matmul(a,b, out=c)
print(c) # content of c has changed

assert c is d # test c and d are the same object not only containing same value
assert id(c), old_id  # make sure that our new c is the same object as the old one

torch.rand(3,2,out=c)
print(c)
assert id(c), old_id
```

```python
# Copying tensor
a= torch.ones(3,4)
b= a.clone()

a[0,1]=900
print(a)
print(b)

assert b is not a # different object in memory
print(torch.eq(a,b))
```

```python
a= torch.rand(2,2, requires_grad=True)
print(a)

b= a.clone()
print(b)

c= a.detach().clone() # We dont get the history of backward process of a tensor, so only the data of the current state itself
print(c)

print(a)
```
