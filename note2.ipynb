# Creating tensors

```python
x= torch.empty(3,4)
print(type(x))
x
```

```python
x= torch.zeros(3,4)
print(x)

x= torch.ones(3,4)
print(x)

torch.manual_seed(669)
x=torch.rand(3,4)
print(x)
```

```python
x= torch.empty(2,3)
print(x, x.shape)

x= torch.empty_like(x)
print(x, x.shape)

x=torch.zeros_like(x)
print(x, x.shape)

x= torch.ones_like(x)
print(x, x.shape)

x=torch.rand_like(x)
print(x, x.shape)
```

```python
some_constant= torch.tensor([3.12,4.2,8.1])
print(some_constant, some_constant.shape)

some_integers= torch.tensor((2,3,4,5,56,5))
print(some_integers, some_integers.shape)
```
